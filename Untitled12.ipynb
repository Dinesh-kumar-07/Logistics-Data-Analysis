```python
import pandas as pd
import os

DATA_PATH = "/content/"

# -----------------------------
# Load datasets
# -----------------------------

orders = pd.read_csv(
    os.path.join(DATA_PATH, "olist_orders_dataset.csv")
)

items = pd.read_csv(
    os.path.join(DATA_PATH, "olist_order_items_dataset.csv")
)

customers = pd.read_csv(
    os.path.join(DATA_PATH, "olist_customers_dataset.csv")
)

sellers = pd.read_csv(
    os.path.join(DATA_PATH, "olist_sellers_dataset.csv")
)

# -----------------------------
# Convert dates
# -----------------------------

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

# -----------------------------
# Calculate item-level totals
# -----------------------------

item_summary = (
    items.groupby("order_id")
    .agg(
        item_count=("order_item_id", "count"),
        order_value=("price", "sum"),
        freight_value=("freight_value", "sum")
    )
    .reset_index()
)

# -----------------------------
# Merge orders + items
# -----------------------------

df = orders.merge(
    item_summary,
    on="order_id",
    how="left"
)

# -----------------------------
# Merge customer information
# -----------------------------

df = df.merge(
    customers[
        [
            "customer_id",
            "customer_unique_id",
            "customer_zip_code_prefix",
            "customer_city",
            "customer_state"
        ]
    ],
    on="customer_id",
    how="left"
)

# -----------------------------
# Delivery lead time
# -----------------------------

df["delivery_lead_time"] = (
    df["order_delivered_customer_date"]
    - df["order_purchase_timestamp"]
).dt.total_seconds() / 86400

# -----------------------------
# Late delivery flag
# -----------------------------

df["late_flag"] = (
    df["order_delivered_customer_date"]
    > df["order_estimated_delivery_date"]
).astype(int)

# -----------------------------
# Month information
# -----------------------------

df["order_month"] = (
    df["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)

df["order_weekday"] = (
    df["order_purchase_timestamp"]
    .dt.day_name()
)

# -----------------------------
# Keep delivered orders
# -----------------------------

df = df[
    df["order_delivered_customer_date"].notna()
].copy()

# Remove impossible negative delivery times

df = df[
    df["delivery_lead_time"] >= 0
]

# -----------------------------
# Save
# -----------------------------

output_path = (
    "/content/olist_order_level_analysis.csv"
)

df.to_csv(
    output_path,
    index=False
)

print("Final dataset shape:", df.shape)

print("\nSample:")
print(df.head())

print("\nColumns:")
print(df.columns.tolist())
```

```python
import pandas as pd

df = pd.read_csv(
    "/content/olist_order_level_analysis.csv"
)

# Total delivered orders
total_orders = len(df)

# Late orders
late_orders = df["late_flag"].sum()

# On-time delivery rate
on_time_rate = (
    1 - df["late_flag"].mean()
) * 100

# Delay rate
delay_rate = (
    df["late_flag"].mean()
) * 100

# Average delivery time
avg_delivery_time = (
    df["delivery_lead_time"].mean()
)

# Average freight
avg_freight = (
    df["freight_value"].mean()
)

# Average order value
avg_order_value = (
    df["order_value"].mean()
)

print("========== LOGISTICS KPIs ==========")

print(
    f"Total Delivered Orders: {total_orders:,}"
)

print(
    f"Late Orders: {late_orders:,}"
)

print(
    f"On-Time Delivery Rate: {on_time_rate:.2f}%"
)

print(
    f"Delay Rate: {delay_rate:.2f}%"
)

print(
    f"Average Delivery Time: "
    f"{avg_delivery_time:.2f} days"
)

print(
    f"Average Freight Cost: "
    f"{avg_freight:.2f}"
)

print(
    f"Average Order Value: "
    f"{avg_order_value:.2f}"
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

df = pd.read_csv(
    "/content/olist_order_level_analysis.csv"
)

os.makedirs(
    "/content/visualizations",
    exist_ok=True
)

# Convert date
df["order_purchase_timestamp"] = pd.to_datetime(
    df["order_purchase_timestamp"]
)

# -----------------------------
# Monthly OTD
# -----------------------------

monthly = (
    df.groupby("order_month")
    .agg(
        delay_rate=("late_flag", "mean")
    )
    .reset_index()
)

monthly["otd_rate"] = (
    1 - monthly["delay_rate"]
) * 100

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly,
    x="order_month",
    y="otd_rate",
    marker="o"
)

plt.title(
    "Monthly On-Time Delivery Rate"
)

plt.xlabel("Month")
plt.ylabel(
    "On-Time Delivery Rate (%)"
)

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "/content/visualizations/monthly_otd.png",
    dpi=300
)

plt.show()
```

```python
# ==========================================
# Visualization 2: Delay Rate by State
# ==========================================

state_delay = (
    df.groupby("customer_state")
    .agg(
        delay_rate=("late_flag", "mean")
    )
    .reset_index()
)

state_delay["delay_rate"] = (
    state_delay["delay_rate"] * 100
)

state_delay = state_delay.sort_values(
    "delay_rate",
    ascending=False
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=state_delay,
    x="delay_rate",
    y="customer_state"
)

plt.title(
    "Delivery Delay Rate by Customer State"
)

plt.xlabel("Delay Rate (%)")
plt.ylabel("Customer State")

plt.tight_layout()

plt.savefig(
    "/content/visualizations/delay_by_state.png",
    dpi=300
)

plt.show()
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ==========================================
# 1. Load dataset
# ==========================================

df = pd.read_csv(
    "/content/olist_order_level_analysis.csv"
)

print("Dataset shape:", df.shape)


# ==========================================
# 2. Select features
# ==========================================

features = [
    "freight_value",
    "order_value",
    "item_count"
]

target = "delivery_lead_time"


# ==========================================
# 3. Remove missing values
# ==========================================

model_data = df.dropna(
    subset=features + [target]
).copy()

print(
    "Rows available for modelling:",
    len(model_data)
)


# ==========================================
# 4. Create X and y
# ==========================================

X = model_data[features]

y = model_data[target]


# ==========================================
# 5. Train-test split
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print(
    "Training rows:",
    len(X_train)
)

print(
    "Testing rows:",
    len(X_test)
)


# ==========================================
# 6. Create model
# ==========================================

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)


# ==========================================
# 7. Train model
# ==========================================

model.fit(
    X_train,
    y_train
)


# ==========================================
# 8. Make predictions
# ==========================================

predictions = model.predict(
    X_test
)


# ==========================================
# 9. Evaluate model
# ==========================================

mae = mean_absolute_error(
    y_test,
    predictions
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        predictions
    )
)

r2 = r2_score(
    y_test,
    predictions
)


# ==========================================
# 10. Display results
# ==========================================

print("\n========== MODEL RESULTS ==========")

print(
    f"MAE  : {mae:.2f} days"
)

print(
    f"RMSE : {rmse:.2f} days"
)

print(
    f"R²   : {r2:.3f}"
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# Actual vs Predicted Delivery Time
# ==========================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x=y_test,
    y=predictions,
    alpha=0.5
)

# Perfect prediction line
minimum = min(
    y_test.min(),
    predictions.min()
)

maximum = max(
    y_test.max(),
    predictions.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.title(
    "Actual vs Predicted Delivery Lead Time"
)

plt.xlabel(
    "Actual Delivery Time (Days)"
)

plt.ylabel(
    "Predicted Delivery Time (Days)"
)

plt.tight_layout()

plt.savefig(
    "/content/visualizations/actual_vs_predicted.png",
    dpi=300
)

plt.show()
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ==========================================
# 1. Load data
# ==========================================

df = pd.read_csv(
    "/content/olist_order_level_analysis.csv"
)

# Convert timestamps

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

for column in date_columns:
    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )


# ==========================================
# 2. Create operational features
# ==========================================

# Seller processing time

df["seller_processing_days"] = (
    df["order_delivered_carrier_date"]
    - df["order_purchase_timestamp"]
).dt.total_seconds() / 86400


# Approval time

df["approval_time_days"] = (
    df["order_approved_at"]
    - df["order_purchase_timestamp"]
).dt.total_seconds() / 86400


# Month

df["purchase_month"] = (
    df["order_purchase_timestamp"]
    .dt.month
)


# Day of week

df["purchase_weekday"] = (
    df["order_purchase_timestamp"]
    .dt.dayofweek
)


# ==========================================
# 3. Select features
# ==========================================

features = [
    "freight_value",
    "order_value",
    "item_count",
    "seller_processing_days",
    "approval_time_days",
    "purchase_month",
    "purchase_weekday"
]

target = "delivery_lead_time"


# ==========================================
# 4. Remove missing values
# ==========================================

model_data = df.dropna(
    subset=features + [target]
).copy()


# Remove impossible values

model_data = model_data[
    (model_data["seller_processing_days"] >= 0) &
    (model_data["approval_time_days"] >= 0) &
    (model_data["delivery_lead_time"] >= 0)
]


print(
    "Rows available:",
    len(model_data)
)


# ==========================================
# 5. Create X and y
# ==========================================

X = model_data[features]

y = model_data[target]


# ==========================================
# 6. Train-test split
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


# ==========================================
# 7. Train model
# ==========================================

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

model.fit(
    X_train,
    y_train
)


# ==========================================
# 8. Predictions
# ==========================================

predictions = model.predict(
    X_test
)


# ==========================================
# 9. Evaluation
# ==========================================

mae = mean_absolute_error(
    y_test,
    predictions
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        predictions
    )
)

r2 = r2_score(
    y_test,
    predictions
)


print("\n========== IMPROVED MODEL ==========")

print(
    f"MAE  : {mae:.2f} days"
)

print(
    f"RMSE : {rmse:.2f} days"
)

print(
    f"R²   : {r2:.3f}"
)


# ==========================================
# 10. Feature Importance
# ==========================================

importance = pd.DataFrame({
    "feature": features,
    "importance": model.feature_importances_
})

importance = importance.sort_values(
    "importance",
    ascending=False
)

print("\nFeature Importance:")
print(importance)
```
